## 03. SQL Data Analysis

### Objective

The objective of this stage is to analyze the cleaned book dataset using MySQL.

### SQL Database Setup

The cleaned dataset generated during the Python preprocessing stage will be imported into MySQL Workbench.

The database will provide a structured environment for performing analytical queries on the 1,000-book dataset.

### Connecting Python with MySQL

```python
import pandas as pd
import mysql.connector

# Load the master cleaned dataset
df_sql = pd.read_csv(
    "../data/cleaned_books.csv",
    encoding="utf-8-sig"
)

print("Dataset shape:", df_sql.shape)
print("Columns:", df_sql.columns.tolist())
```

```python
conn = mysql.connector.connect(
    host="127.0.0.1",
    port=3306,
    user="YOUR_MYSQL_USERNAME",
    password="YOUR_MYSQL_PASSWORD",
    database="bookwise_analytics",
    connection_timeout=5
)

cursor = conn.cursor()

print("MySQL connection successful.")
```

### Loading Data into MySQL

```python
# Prepare column names for MySQL
df_sql = df_sql.rename(columns={
    "Product Type": "Product_Type",
    "Number Available": "Number_Available",
    "Number of Reviews": "Number_of_Reviews",
    "Rating Category": "Rating_Category",
    "Price Band": "Price_Band",
    "Stock Status": "Stock_Status",
    "Title Length": "Title_Length",
    "Description Length": "Description_Length",
    "Description Word Count": "Description_Word_Count"
})

columns = [
    "Title",
    "Price",
    "Rating",
    "Availability",
    "Link",
    "Category",
    "Description",
    "UPC",
    "Product_Type",
    "Tax",
    "Number_Available",
    "Number_of_Reviews",
    "Stock_Quantity",
    "Rating_Category",
    "Price_Band",
    "Stock_Status",
    "Title_Length",
    "Description_Length",
    "Description_Word_Count"
]

# Convert NaN to None for MySQL
data = []

for _, row in df_sql.iterrows():
    values = []

    for column in columns:
        value = row[column]

        if pd.isna(value):
            value = None

        values.append(value)

    data.append(tuple(values))

print("Records prepared for MySQL:", len(data))
```

```python
insert_query = """
INSERT INTO books (
    Title,
    Price,
    Rating,
    Availability,
    Link,
    Category,
    Description,
    UPC,
    Product_Type,
    Tax,
    Number_Available,
    Number_of_Reviews,
    Stock_Quantity,
    Rating_Category,
    Price_Band,
    Stock_Status,
    Title_Length,
    Description_Length,
    Description_Word_Count
)
VALUES (
    %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,
    %s, %s, %s, %s, %s, %s, %s, %s, %s
)
"""

cursor.executemany(insert_query, data)
conn.commit()

print("Data imported successfully.")
print("Records inserted:", cursor.rowcount)
```
